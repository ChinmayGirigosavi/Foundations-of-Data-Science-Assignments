**Q1.** Write a Python program to find all null values in a given dataset and remove them.

**Objective:** To identify missing values in a dataset and remove rows containing null values.

````python

import pandas as pd

# Download dataset from GitHub
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

# Read the dataset
data = pd.read_csv(url)

# Display original dataset
print("Original Dataset:")
display(data.head())

# Find and display null values
print("\nNull values in each column:")
print(data.isnull().sum())

# Remove rows containing null values
clean_data = data.dropna()

# Display cleaned dataset
print("\nDataset after removing null values:")
display(clean_data.head())

# Check null values after removal
print("\nNull values after removal:")
print(clean_data.isnull().sum())

# Save cleaned dataset
clean_data.to_csv("titanic_cleaned.csv", index=False)

print("\nCleaned dataset saved as titanic_cleaned.csv")
````

**Q2.** Write a Python program to convert categorical values into numeric format for a given dataset.

**Objective:** To convert categorical data into numerical values using category codes.

````python
import pandas as pd
from sklearn.datasets import load_iris

# Load the built-in Iris dataset
iris = load_iris()

# Create DataFrame
data = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Add categorical species column
data["species"] = pd.Categorical.from_codes(
    iris.target,
    iris.target_names
)

# Display original dataset
print("Original Dataset:")
display(data.head())

# Convert categorical values into numeric values
data["species_numeric"] = data["species"].astype("category").cat.codes

# Display converted dataset
print("\nDataset after converting categorical values to numeric:")
display(data.head())

# Display mapping
print("\nCategorical to Numeric Mapping:")
mapping = dict(enumerate(data["species"].astype("category").cat.categories))
print(mapping)
````

**Q1.** Write a Python program to rescale the data between 0 and 1 using an inbuilt dataset.

**Objective:** To normalize the dataset between 0 and 1 using MinMaxScaler.

````python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler

# Load inbuilt Iris dataset
iris = load_iris()

# Convert dataset into DataFrame
data = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Display original data
print("Original Dataset:")
display(data.head())

# Create MinMaxScaler
scaler = MinMaxScaler()

# Rescale data between 0 and 1
scaled_data = scaler.fit_transform(data)

# Convert scaled data into DataFrame
scaled_data = pd.DataFrame(
    scaled_data,
    columns=data.columns
)

# Display rescaled data
print("\nRescaled Dataset:")
display(scaled_data.head())

# Check minimum and maximum values
print("\nMinimum values:")
print(scaled_data.min())

print("\nMaximum values:")
print(scaled_data.max())
````

**Step 4:** Observe the output.

**You will see:**
- Original Iris dataset.
- Rescaled dataset.
- Minimum values.
- Maximum values.

**Expected result:**
- Minimum values should be 0.0.
- Maximum values should be 1.0.

**Q2.** Write a Python program to split the dataset into training and testing sets.

**Objective:** To divide a dataset into training data and testing data using train_test_split().

````python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

# Load the inbuilt Iris dataset
iris = load_iris()

# Create features and target
X = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

y = iris.target

# Display original dataset
print("Original Dataset:")
display(X.head())

# Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Display results
print("\nTraining data size:", len(X_train))
print("Testing data size:", len(X_test))

print("\nTraining Data:")
display(X_train.head())

print("\nTesting Data:")
display(X_test.head())
````

**Q1.** Write a Python program to implement complete data preprocessing in a given dataset.

**Objective:** To perform complete data preprocessing using the following operations:
1. Handling missing values.
2. Encoding categorical values.
3. Splitting the dataset into training and testing sets.
4. Feature scaling.

````python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# 1. Download dataset from GitHub
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
data = pd.read_csv(url)

print("Original Dataset:")
display(data.head())

# 2. Check missing values
print("\nMissing values:")
print(data.isnull().sum())

# 3. Handle missing values
data["Age"] = data["Age"].fillna(data["Age"].median())
data["Embarked"] = data["Embarked"].fillna(data["Embarked"].mode()[0])
data["Cabin"] = data["Cabin"].fillna("Unknown")

print("\nMissing values after handling:")
print(data.isnull().sum())

# 4. Remove unnecessary columns
data = data.drop(
    ["PassengerId", "Name", "Ticket", "Cabin"],
    axis=1
)

# 5. Encode categorical values
data = pd.get_dummies(
    data,
    columns=["Sex", "Embarked"],
    drop_first=True
)

print("\nDataset after encoding:")
display(data.head())

# 6. Separate features and target
X = data.drop("Survived", axis=1)
y = data["Survived"]

# 7. Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining data size:", X_train.shape)
print("Testing data size:", X_test.shape)

# 8. Feature scaling
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# 9. Display scaled data
print("\nScaled Training Data:")
print(X_train[:5])

print("\nScaled Testing Data:")
print(X_test[:5])

print("\nData preprocessing completed successfully!")
````
