SET A – Question 1: Titanic Dataset
Question: Load the Titanic dataset, show basic information, check for missing values, calculate basic statistics, and impute missing values for the Age column.
Step 1: Import required libraries

```python
import pandas as pd
import numpy as np
import seaborn as sns
```

Step 2: Load the Titanic dataset

```python
data = sns.load_dataset("titanic")

print("First five records:")
print(data.head())
```

Step 3: Display basic information

```python
print("Dataset Information:")
print(data.info())

print("\nDataset Shape:")
print(data.shape)

print("\nColumn Names:")
print(data.columns.tolist())
```

Step 4: Check missing values

```python
print("Missing Values:")
print(data.isnull().sum())
```

Step 5: Calculate basic statistics

```python
print("Basic Statistical Information:")
print(data.describe())
```

Step 6: Find missing values in Age

```python
print("Missing Age Values:")
print(data['age'].isnull().sum())
```

Step 7: Fill missing Age values
We will use the median of the Age column to fill missing values.

```python
median_age = data['age'].median()

data['age'] = data['age'].fillna(median_age)

print("Missing Age Values After Imputation:")
print(data['age'].isnull().sum())
```

Step 8: Display the cleaned dataset

```python
print("Cleaned Dataset:")
print(data.head())

print("\nMissing Values After Cleaning:")
print(data.isnull().sum())
```

Result: The Titanic dataset was loaded successfully. Basic information and statistics were displayed, missing values were checked, and missing Age values were replaced with the median.


SET A – Question 2: Titanic Data Visualization
Question: Produce the following visualizations using the Titanic dataset:
1. Bar plot for sex.
2. Histogram for age.
3. Boxplot for fare.
4. Scatter plot for age versus fare.
5. Heatmap of the correlation matrix.
Step 1: Import required libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Step 2: Load the Titanic dataset

```python
data = sns.load_dataset("titanic")

print(data.head())
```

Step 3: Bar plot for sex

```python
sns.countplot(x='sex', data=data)

plt.title("Bar Plot of Sex")
plt.xlabel("Sex")
plt.ylabel("Count")

plt.show()
```

Step 4: Histogram for age

```python
sns.histplot(data['age'].dropna(), bins=20, kde=True)

plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

Step 5: Boxplot for fare

```python
sns.boxplot(x=data['fare'])

plt.title("Boxplot of Fare")
plt.xlabel("Fare")

plt.show()
```

Step 6: Scatter plot for age versus fare

```python
sns.scatterplot(
    x='age',
    y='fare',
    data=data
)

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```

Step 7: Heatmap of correlation matrix

```python
correlation = data.select_dtypes(include='number').corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()
```

Result: Five visualizations were successfully created using the Titanic dataset with Seaborn and Matplotlib.

SET B – Question 1: Red Wine Quality Dataset
Question: Show the distribution of wine quality, calculate skewness and kurtosis, and apply a log transformation to the alcohol column.
Step 1: Import required libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Step 2: Load the red wine dataset

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"

data = pd.read_csv(url, sep=';')

print("First five records:")
print(data.head())
```

Step 3: Display dataset information

```python
print("Dataset Information:")
print(data.info())

print("\nDataset Shape:")
print(data.shape)
```

Step 4: Display wine quality distribution

```python
print("Wine Quality Distribution:")
print(data['quality'].value_counts().sort_index())
```

Step 5: Plot wine quality distribution

```python
sns.countplot(x='quality', data=data)

plt.title("Wine Quality Distribution")
plt.xlabel("Quality")
plt.ylabel("Count")

plt.show()
```

Step 6: Calculate skewness

```python
skewness = data['alcohol'].skew()

print("Skewness of Alcohol:", skewness)
```

Step 7: Calculate kurtosis

```python
kurtosis = data['alcohol'].kurtosis()

print("Kurtosis of Alcohol:", kurtosis)
```

Step 8: Apply log transformation
We will use np.log() because the alcohol values are positive.

```python
data['alcohol_log'] = np.log(data['alcohol'])

print("Original Alcohol Values:")
print(data['alcohol'].head())

print("\nLog Transformed Alcohol Values:")
print(data['alcohol_log'].head())
```

Step 9: Visualize original and transformed alcohol

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.hist(data['alcohol'], bins=20)
plt.title("Original Alcohol")
plt.xlabel("Alcohol")

plt.subplot(1, 2, 2)
plt.hist(data['alcohol_log'], bins=20)
plt.title("Log Transformed Alcohol")
plt.xlabel("Log Alcohol")

plt.tight_layout()
plt.show()
```

Result: The red wine quality distribution was displayed, skewness and kurtosis were calculated, and a log transformation was applied to the alcohol column.

SET B – Question 2: Penguins Dataset Visualization
Question: Create visualizations using Matplotlib (line plot), Seaborn (pairplot), and Plotly (interactive scatter plot) using the Penguins dataset.
Step 1: Import required libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

Step 2: Load the Penguins dataset

```python
data = sns.load_dataset("penguins")

print("First five records:")
print(data.head())
```

Step 3: Remove missing values for visualization

```python
data = data.dropna()

print("Cleaned Dataset:")
print(data.head())

print("\nDataset Shape:")
print(data.shape)
```

Step 4: Create a line plot using Matplotlib
This plot shows the average body mass of each penguin species.

```python
average_mass = data.groupby('species')['body_mass_g'].mean()

plt.plot(
    average_mass.index,
    average_mass.values,
    marker='o'
)

plt.title("Average Body Mass of Penguins")
plt.xlabel("Species")
plt.ylabel("Average Body Mass (g)")

plt.show()
```

Step 5: Create a pairplot using Seaborn

```python
sns.pairplot(
    data,
    hue='species'
)

plt.show()
```

Step 6: Create an interactive scatter plot using Plotly

```python
fig = px.scatter(
    data,
    x='flipper_length_mm',
    y='body_mass_g',
    color='species',
    hover_data=['island', 'sex'],
    title="Penguins Interactive Scatter Plot"
)

fig.show()
```

Result: The Penguins dataset was visualized using Matplotlib, Seaborn, and Plotly. An interactive scatter plot was successfully created.

SET C – Question 1: Interactive Earthquake Map using Folium
Question: Create an interactive map using Folium, where the location of each earthquake is marked and color-coded based on its magnitude.
Step 1: Install and import Folium

```python
!pip install folium
```

```python
# Now import the libraries:

import folium
import requests
```

Step 2: Load earthquake dataset
We will use the USGS earthquake GeoJSON feed, which provides recent earthquake information.

```python
url = "https://earthquake.usgs.gov/earthquakes/feed/v1.0/summary/all_month.geojson"

response = requests.get(url)
earthquake_data = response.json()

print("Earthquake Dataset Loaded Successfully!")
print("Total Earthquakes:", len(earthquake_data['features']))
```

Step 3: Create the map

```python
earthquake_map = folium.Map(
    location=[0, 0],
    zoom_start=2,
)

print("Earthquake Map Created Successfully!")
```

Step 4: Define colors based on magnitude

```python
def get_color(magnitude):

    if magnitude < 2:
        return 'green'

    elif magnitude < 4:
        return 'orange'

    elif magnitude < 5:
        return 'red'

    else:
        return 'darkred'
```

Step 5: Add earthquake locations to the map

```python
for earthquake in earthquake_data['features']:

    properties = earthquake['properties']
    coordinates = earthquake['geometry']['coordinates']

    longitude = coordinates[0]
    latitude = coordinates[1]
    magnitude = properties['mag']

    if magnitude is None:
        continue

    color = get_color(magnitude)

    folium.CircleMarker(
        location=[latitude, longitude],
        radius=max(3, magnitude * 1.5),
        popup=f"Magnitude: {magnitude}<br>Place: {properties['place']}",
        color=color,
        fill=True,
        fill_color=color,
        fill_opacity=0.7
    ).add_to(earthquake_map)
```

Step 6: Display the interactive map

```python
earthquake_map
```

Step 7: Save the map as an HTML file

```python
earthquake_map.save("earthquake_map.html")

print("Earthquake Map Saved Successfully!")
```

Result: An interactive earthquake map was successfully created using Folium. Earthquake locations are displayed with markers, and marker colors represent their magnitudes.
